```python
import random
import matplotlib.pyplot as plt
import numpy as np
import math
from scipy.special import comb
import scipy.stats as st
import networkx as nx
```

```python
def powerlaw_sample(alpha):
    xmin=1
    u=random.uniform(0,1)
    return xmin*math.pow(u, 1/(1-alpha))
```

```python
#幂律分布概率密度
def powerlaw(x):
    xmin=1
    alpha=3
    return ((alpha-1)/xmin)*np.power((x/xmin),-alpha)
plt.figure(figsize=(8,6))
x=np.arange(1,10,0.1)
y=powerlaw(x)
plt.scatter(x,y)
plt.xscale('log')
plt.yscale('log')
plt.show()
```

```python
#验证幂律分布
sample_number=100000
alpha=2.1
test_list=[]
for i in range(sample_number):
    test_list.append(powerlaw_sample(alpha))
print(np.mean(test_list))
```
